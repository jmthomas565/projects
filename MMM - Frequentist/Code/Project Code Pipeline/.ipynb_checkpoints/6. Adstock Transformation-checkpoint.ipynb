```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
```

```python
#Defining adstock function
def adstock(x, theta):
    return tsa.filters.recursive_filter(x, theta)
```

```python

#Cross-validation hold-out
hold_out_split = -26
y_train_calib = y_train.iloc[:hold_out_split]
y_train_hold_out = y_train.iloc[hold_out_split:]


#Defining adj_R^2 function
def adj_r2(y_true, y_pred, p):
    r2 = r2_score(y_true, y_pred)
    n = len(y_true)
    #Making sure the denominator of adj R^2 > 0
    if n - p - 1 <= 0:
        return np.nan
    calculation = 1 - (1 - r2) * (n - 1) / (n - p - 1)
    return calculation
```

```python
#Testing online channels across the adstock parameter space, whilst holding offline channels fixed
def test_adstock_models_adj_r2(X_raw, adstock_decay_values, adstock_cols, other_cols, feature_sets):
    for decay in adstock_decay_values:
        #creates a copy of X_raw to avoid a cumulative adstock effect
        X_copy = X_raw.copy()
        
        #adstock is applied to the copied column
        for col in adstock_cols:
            X_copy[f"{col}_adstock"] = adstock(X_copy[col], [decay])
        
        print(f"Decay Value: {round(decay, 2)}")
        
        #Testing each subset of regressors
        for name, features_to_use in feature_sets.items():
            X_small = X_copy[features_to_use]
            
            #splitting for hold-out cv
            X_train_calib = X_small.iloc[:hold_out_split]
            X_val_hold_out = X_small.iloc[hold_out_split:]
            
            #training the ridge reg model 
            ridge = Ridge(alpha=1, positive = True)
            ridge.fit(X_train_calib, y_train_calib)
            y_pred = ridge.predict(X_val_hold_out)
            
            score = adj_r2(y_train_hold_out, y_pred, p=X_small.shape[1])
            
            print(f"{name}: Adjusted R2 = {round(score, 4)}")

#Setting the parameter space
adstock_decay_values = np.arange(0.1, 1.0, 0.1)
#Defining the variables to be tested over
adstock_columns = ["branded_search_spend", "nonbranded_search_spend", "facebook_spend"]
other_columns = ['print_spend', 'ooh_spend', 'tv_spend', 'radio_spend']

#This sets up the sets of regressors, noting that there are 3 online channels (adstock_columns), and 4 offline channels (other columns)
regressor_set = {
    "4 regressors": [adstock_columns[0], adstock_columns[1], adstock_columns[2], other_columns[0]],
    "5 regressors": [adstock_columns[0], adstock_columns[1], adstock_columns[2], other_columns[0], other_columns[1]],
    "6 regressors": [adstock_columns[0], adstock_columns[1], adstock_columns[2], other_columns[0], other_columns[1], other_columns[2]],
    "7 regressors": [adstock_columns[0], adstock_columns[1], adstock_columns[2], other_columns[0], other_columns[1], other_columns[2], other_columns[3]]
}

adstocked_regressor_sets = {
    "4 regressors": [f"{col}_adstock" for col in adstock_columns] + [other_columns[0]],
    "5 regressors": [f"{col}_adstock" for col in adstock_columns] + other_columns[:2],
    "6 regressors": [f"{col}_adstock" for col in adstock_columns] + other_columns[:3],
    "7 regressors": [f"{col}_adstock" for col in adstock_columns] + other_columns[:4]
}
# Now call the combined function
test_adstock_models_adj_r2(X_train, adstock_decay_values, adstock_columns, other_columns, adstocked_regressor_sets)
```
