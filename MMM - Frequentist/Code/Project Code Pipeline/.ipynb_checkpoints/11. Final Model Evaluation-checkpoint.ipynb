```python
#making the final changes to X_train

def adstock(x, theta):
    return tsa.filters.recursive_filter(x, theta)

#locking in adstock decay rate for branded_search_spend (the only channel variable needed for the model)
X_train["branded_search_spend_adstock"] = adstock(X_train["branded_search_spend"], [0.8])

X_train['sales_lag_1'] = y_train['sales'].shift(-1)

#selecting the final cols which performed best in the final cross-valiation
X_train = X_train[[
    'sales_lag_1',
    'branded_search_spend_adstock'
]]

X_train = X_train.iloc[:-1]

y_train = y_train.iloc[:-1]
```

```python

#Then doing the same for X_test

X_test["branded_search_spend_adstock"] = adstock(X_test["branded_search_spend"], [0.8])

X_test['sales_lag_1'] = y_test['sales'].shift(-1)


X_test = X_test[[
    'sales_lag_1',
    'branded_search_spend_adstock'
]]

X_test = X_test.iloc[:-1]

y_test = y_test.iloc[:-1]
```

```python
#Calculating adj_R^2

def adj_r2(y_true, y_pred, p):
    r2 = r2_score(y_true, y_pred)
    n = len(y_true)
    #Making sure the denominator of adj R^2 > 0
    if n - p - 1 <= 0:
        return np.nan
    calculation = 1 - (1 - r2) * (n - 1) / (n - p - 1)
    return calculation
```

```python
#testing the final model, using X_test and y_test

import pandas as pd
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score


def test_model_adj_r2(X_regressors, regressors):

    ridge = Ridge(alpha=10, positive = True)
    ridge.fit(X_regressors, y_train)
    
    y_pred = ridge.predict(X_test)
    
    score = adj_r2(y_test, y_pred, p=X_regressors.shape[1])
    
    #calculating MAPE
    mape = np.mean(np.abs((y_test - y_pred) / y_test)) * 100

    #Printing the final adj_R^2 values and MAPE           
    print(f"{regressors} Full model: Adjusted R² = {round(score, 4)}, MAPE = {round(mape, 2)}%")


#Fitting the final evaluated model
ridge = Ridge(alpha=10, positive = True)
ridge.fit(X_train, y_train)

y_pred = ridge.predict(X_test)

score = adj_r2(y_test, y_pred, p=X_train.shape[1])
mape = np.mean(np.abs((y_test - y_pred) / y_test)) * 100

#Printing the final adj_R^2 values            
print(f"Adjusted R² = {round(score, 4)}")
print(f"MAPE = {round(mape, 4)}")

X_2reg = X_train.iloc[:, :2]
test_model_adj_r2(X_2reg, "2 regressors")
```
