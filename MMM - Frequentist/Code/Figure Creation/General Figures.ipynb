```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

### Geometric Decay

```python

#creating plot to visualise geometric decay 

days = np.arange(0, 11)

decay = 10 * np.exp(-0.5 * days)

fig, ax = plt.subplots(figsize=(10, 6))

#Using appropriate font
plt.rcParams['font.family'] = 'Times New Roman'

bars = ax.bar(days, decay, width=1, edgecolor='black', linewidth=1)

#Setting only the first bar as red, and all others as grey

bars[0].set_facecolor('red')
for bar in bars[1:]:
    bar.set_facecolor('gray')


ax.set_xticks(range(0, 11))
ax.set_yticks(range(0, 12))


ax.xaxis.set_tick_params(labelsize=10)
ax.yaxis.set_tick_params(labelsize=10)


#Removing the top and right spines

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

#setting label size 

ax.tick_params(labelsize=12)

#setting background colour (axis level) 

ax.set_facecolor('#fafafa')

ax.grid ( color='grey', linestyle='-', linewidth=0.25, alpha=0.5)

#editing the grid

ax.grid(True, alpha=0.3)

ax.set_xlabel('Days', fontsize=12)


#removing the spines (top and right side)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

for bar in bars:
    bar.set_zorder(3)

ax.plot(days, decay, 'k-', linewidth=2, zorder=4)

plt.tight_layout()
plt.show()
```

### Visualisation of the Hill Function

```python
def saturation(x, K, S, epsilon=1e-6):
    K = max(K, epsilon)
    return 1 / (1 + (x / K) ** -S)
```

```python
X = np.linspace(0.01, 5, 1000)
```

```python
s = 2.5

K1 = saturation(X, 0.5, s)
K2 = saturation(X, 0.7, s)
K3 = saturation(X, 0.9, s)
K4 = saturation(X, 1.1, s)
K5 = saturation(X, 1.3, s)

K = 0.9

s1 = saturation(X, K, 1)
s2 = saturation(X, K, 1.5)
s3 = saturation(X, K, 2)
s4 = saturation(X, K, 2.5)
s5 = saturation(X, K, 3)
```

```python
import matplotlib as mpl
print(mpl.matplotlib_fname())
```

```python

fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout = True, figsize = (10, 6))

#Using appropriate font
plt.rcParams['font.family'] = 'Times New Roman'


ax1.plot(X, K1, label = 0.5)
ax1.plot(X, K2, label =0.7)
ax1.plot(X, K3, label = 0.9)
ax1.plot(X, K4, label = 1.1)
ax1.plot(X, K5, label = 1.3)


#removing the spines from top and right sides
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

#setting the label padding
ax1.xaxis.labelpad=20
ax1.yaxis.labelpad=20

#setting the background colour (at axis level)
ax1.set_facecolor('#fafafa')

#setting the grids across the axis
ax1.grid (color='grey', linestyle='-', linewidth=0.25, alpha=0.5)

ax1.set_title("K changes, while s = 2.5")

ax1.legend(title = "K", loc = 'lower right')


ax2.plot(X, s1, label = 1)
ax2.plot(X, s2, label = 1.5)
ax2.plot(X, s3, label = 2)
ax2.plot(X, s4, label = 2.5)
ax2.plot(X, s5, label = 3)


ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)



ax2.xaxis.labelpad=20
ax2.yaxis.labelpad=20

ax2.set_facecolor('#fafafa')

ax2.grid (color='grey', linestyle='-', linewidth=0.25, alpha=0.5)

ax2.set_title("s changes, while K = 0.9")


ax2.legend(title = "s", loc = 'lower right')
```
