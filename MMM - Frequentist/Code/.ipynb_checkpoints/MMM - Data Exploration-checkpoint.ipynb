### Printing Simple Descriptive Stats

```python
import warnings


with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    desciptive_stats_biks_v1 = bike_sales_data[['sales', 'branded_search_spend', 'nonbranded_search_spend',
           'facebook_spend']].agg([min, max, 'mean', 'var'])


desciptive_stats_biks_v1.round()
```

```python
with warnings.catch_warnings():
    warnings.simplefilter('ignore')

    desciptive_stats_biks_v2 = bike_sales_data[['print_spend', 'ooh_spend', 'tv_spend',
           'radio_spend']].agg([min, max, 'mean', 'var'])

desciptive_stats_biks_v2.round()
```

```python
def macro_data_aggs(agg_col):
    df = agg_col.agg([min, max, 'mean', 'var'])
    return df.to_frame()
```

```python
base_rate_usa_desciptive_stats = macro_data_aggs(base_rate_usa['FEDFUNDS'])

gdp_usa_desciptive_stats = macro_data_aggs(gdp_usa['NA000334Q'])

unemployment_rate_usa_desciptive_stats = macro_data_aggs(unemployment_rate_usa['UNRATENSA'])
```
