```python
# Imports
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt 
import matplotlib.dates as mdates
```

```python
bike_sales_data = pd.read_csv("bike_sales_data.csv")

base_rate_usa = pd.read_csv("base_rate_usa.csv")

gdp_usa = pd.read_csv("gdp_usa.csv")

unemployment_rate_usa = pd.read_csv("unemployment_rate_usa.csv")
```

### Setting Indexes

```python
def index_to_date(df, col):
    
    df = df.set_index(col) 
    df.index = pd.to_datetime(df.index)
    return(df)
```

```python
bike_sales_data = index_to_date(bike_sales_data, 'Week')

base_rate_usa = index_to_date(base_rate_usa, 'observation_date')

gdp_usa = index_to_date(gdp_usa, 'observation_date')

unemployment_rate_usa = index_to_date(unemployment_rate_usa, 'observation_date')
```

### Filtering by Relevant Dates

```python
bike_sales_data.index.min()
```

```python
bike_sales_data.index.max()
```

```python
def date_filter(df):

    df = df.loc['2017-07-23':'2022-07-10']
    return df
```

```python
base_rate_usa = date_filter(base_rate_usa)

gdp_usa = date_filter(gdp_usa)

unemployment_rate_usa = date_filter(unemployment_rate_usa)
```

```python
def macro_data_aggs(agg_col):
    df = agg_col.agg([min, max, 'mean', 'var'])
    return df.to_frame()
```

```python
base_rate_usa_desciptive_stats = macro_data_aggs(base_rate_usa['FEDFUNDS'])

gdp_usa_desciptive_stats = macro_data_aggs(gdp_usa['NA000334Q'])

unemployment_rate_usa_desciptive_stats = macro_data_aggs(unemployment_rate_usa['UNRATENSA'])
```
